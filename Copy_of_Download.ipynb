```python
gdp_df = gdp_tables[2].copy()
gdp_df = gdp_df.dropna(how='all')
gdp_df.head(10)
```

```python
gdp_df['Country/Territory'] = gdp_df['Country/Territory'].str.replace(r'\[.*?\]', '', regex=True).str.strip()
gdp_df.head(10)
```

```python
merged_df = pd.merge(df, gdp_df, left_on='Country or territory', right_on='Country/Territory', how='inner')
merged_df.head(10)
```

```python
merged_df['IMF (2026)[1]'] = merged_df['IMF (2026)[1]'].astype(str).str.replace(',', '').str.extract(r'(\d+)').astype(float)
```

```python
correlation = merged_df['Population (1 July 2023)'].corr(merged_df['IMF (2026)[1]'])
print(f"Correlation between Population and GDP: {correlation}")
```

```python
import matplotlib.pyplot as plt

top10 = merged_df.nlargest(10, 'Population (1 July 2023)')
plt.figure(figsize=(10,6))
plt.barh(top10['Country or territory'], top10['Population (1 July 2023)'])
plt.xlabel('Population')
plt.title('Top 10 Most Populated Countries')
plt.gca().invert_yaxis()  # highest population on top
plt.show()
```

```python
top10 = merged_df[merged_df['Country or territory'] != 'World'].nlargest(10, 'Population (1 July 2023)')

plt.figure(figsize=(10,6))
plt.barh(top10['Country or territory'], top10['Population (1 July 2023)'])
plt.xlabel('Population')
plt.title('Top 10 Most Populated Countries')
plt.gca().invert_yaxis()
plt.show()
```
